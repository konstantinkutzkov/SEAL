```python
import numpy as np
from collections import defaultdict, Counter
import sys
sys.path.append("../src")
from utils import *
```

```python
def get_attribute_frequencies(datapath, filename):
    entity_types_map = {} 
    all_types = set()
    f = open(datapath+filename, 'r', encoding='utf8')
    for line in f:
        ls = line.split('^^')
        entity = ls[0].split()[0]
        entity = entity.split('/')[-1][:-1]
        entity_types_map.setdefault(entity, {'string': 0})
        if len(ls) > 1:
            tp = ls[-1][1:-4]
            tp = tp.split('#')[-1]
            entity_types_map[entity].setdefault(tp, 0)
            entity_types_map[entity][tp] += 1
            all_types.add(tp)
        else:
            entity_types_map[entity]['string'] += 1
    f.close()
    return entity_types_map, all_types

prefix = 'fr_'
datapath = f'../data/{prefix}en/'

attr_map_en, attr_set_en = get_attribute_frequencies(datapath, 'en_att_triples')
attr_map_fr, attr_set_fr = get_attribute_frequencies(datapath, 'fr_att_triples')
```
